```python
import os
from IPython.display import display, Markdown
from openai import OpenAI
from playwright.sync_api import sync_playwright
from bs4 import BeautifulSoup
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
GEMINI_BASE_URL = 'https://generativelanguage.googleapis.com/v1beta/openai/'
GEMINI_API_KEY = os.getenv('GOOGLE_API_KEY')
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=GEMINI_API_KEY)
```

```python
OLLAMA_BASE_URL = 'http://localhost:11434/v1'
OLLAMA_API_KEY = 'ollama' # can be any words, not important
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key=OLLAMA_API_KEY)
```

```python
system_prompt = """
You are a helpful education coach to help to design a professional 4 weeks plan for the user based on the job title. 
If the job title is not valid, please say so and ask user to input the correct job title again. 
If the job title is valid, Response in markdown. Do not wrap the markdown with code block - reply only in markdown.
"""
user_prompt = """
You will analyze the job title provided by user, think thoroughly before providing a professional plan regarding the structures of the learning path, provide useful link for the learning & tips to achieve success for the user.
This is the job title user wish to expertise on:

"""
```

```python
job = "machine learning engineer"

message = [
    {'role': 'system', 'content': system_prompt},
    {'role': 'user', 'content': user_prompt + job}
]
```

```python
#GEMINI

gemini_response = gemini.chat.completions.create(model='gemini-3.1-flash-lite-preview', messages=message)
result = gemini_response.choices[0].message.content
display(Markdown(result))
```

```python
#OLLAMA

ollama_response = ollama.chat.completions.create(model='gpt-oss:latest', messages=message)
result = ollama_response.choices[0].message.content
display(Markdown(result))
```
